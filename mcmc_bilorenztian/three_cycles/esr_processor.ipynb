```python
# Compile regular expressions
time_stamp = re.compile('202\\d-\\d{2}-\\d{2}_\\d*h\\d*m\\d*s')


# Functions to extract attributes
def extract_attributes(filename):
    a = time_stamp.findall(f[0])[0].split('_')
    date_str, time_str = a[0], a[1]
    year, month, day = date_str.split('-')
    time_parts = re.findall(r'\d+', time_str)
    h, m, s = time_parts
    mt = time.mktime(datetime.datetime(int(year), int(month), int(day), int(h), int(m), int(s)).timetuple())   
    return mt

# File filtering function
def filter_files(file_list):
    filtered_files = sorted([f for f in file_list], key=lambda x: extract_attributes(x)[0])
    return filtered_files 
```
